## TASK 1 Discount Checking
1. If the Price is 5000 or more, give a 25% discount.
2. If the Price is 3000 or more, give a 15% discount.
3. If the Price is 1000 or more, give a 10% discount.
4. If the Price is less than 1000, give no discount.
5. Calculate and display the discount Price.
6. Calculate and display the final Price. after discount

```python
Price = float(input("Enter shopping Price: "))

if Price >= 5000:
    discount = 25
elif Price >= 3000:
    discount = 15
elif Price >= 1000:
    discount = 10
else:
    discount = 0

discount_Price = Price * discount / 100
final_Price = Price - discount_Price

print("Discount:", discount, "%")
print("Discount Price:", discount_Price)
print("Final Price:", final_Price)
```

## Task 3 — Number Checker
1. Whether the number is positive or negative.
2. Whether the number is even or odd.
3. Whether the number is greater than 10 and less than 50.
4. Whether the number is divisible by 5 or 10.

```python
num = int(input("Enter a number: "))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")

if num % 2 == 0:
    print("Even")
else:
    print("Odd")

if num > 10 and num < 50:
    print("Number is between 10 and 50")
else:
    print("Number is not between 10 and 50")

if num % 5 == 0 or num % 10 == 0:
    print("Number is divisible by 5 or 10")
else:
    print("Number is not divisible by 5 or 10")
```
